```python
plt.figure(figsize=(10,5))
df["restaurant type"].value_counts().head(10).plot(kind="bar")

plt.title("Top 10 restaurant Types ")
```

```python
plt.figure(figsize=(8,5))

plt.hist(df["rate (out of 5)"].dropna(), bins=20)

plt.title("Distribution of Restaurant Ratings")
plt.xlabel("Rating")
plt.ylabel("Number of Restaurants")

plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.hist(df["avg cost (two people)"].dropna(), bins=30)

plt.title("Distribution of Average Cost for Two People")
plt.xlabel("Average Cost")
plt.ylabel("Number of Restaurants")

plt.show()
```

```python
plt.figure(figsize=(6,5))

df["online_order"].value_counts().plot(kind="bar")

plt.title("Restaurants Offering Online Order")
plt.xlabel("Online Order")
plt.ylabel("Number of Restaurants")

plt.show()
```

```python
top_areas = df["area"].value_counts().head(10)

plt.figure(figsize=(10,5))

top_areas.plot(kind="bar")

plt.title("Top 10 Areas with Most Restaurants")
plt.xlabel("Area")
plt.ylabel("Number of Restaurants")
plt.xticks(rotation=45)

plt.show()
```

```python
top_cuisines = df["cuisines type"].value_counts().head(10)

plt.figure(figsize=(10,5))

top_cuisines.plot(kind="bar")

plt.title("Top 10 Popular Cuisines")
plt.xlabel("Cuisine")
plt.ylabel("Number of Restaurants")
plt.xticks(rotation=45)

plt.show()
```
