TASK 3.3 NumPy Vectorized Operations

Generate two random integers Arrays of size 10 with values between 1 and 20

1. Perform element wise addition,subtraction, multiplication and division between the arrays.

```python
arr2=np.random.randint(1,20,size=10)

arr3=np.random.randint(1,20,size=(10))
```

```python
print(arr2)
```

```python
print(arr3)
```

```python
print("Addition: ",arr2 +arr3)
```

```python
print("Subtraction: ",arr2 -arr3)
```

```python
print("Multiplication: ",arr2 *arr3)
```

```python
print("Division: ",arr2 /arr3)
```

2. Square every element in the first array

```python
print("Square: ",arr2**2)
```

3. Dot product of the two arrays

```python
print("Dot product: ",np.dot(arr2,arr3))
```

4. Sort the second array in Ascending order.

```python
print("arr3 in Ascending order: ",np.sort(arr3))
```
